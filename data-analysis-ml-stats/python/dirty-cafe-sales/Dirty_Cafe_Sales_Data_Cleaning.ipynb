# Data Cleaning Project: Dirty Cafe Sales


## 1. Import Libraries
We import pandas for data manipulation and numpy for missing values.

```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
```

## 2. Load Dataset

```python
df = pd.read_csv('dirty_cafe_sales.csv')
df.head()
```

3. Explore the Dataset

```python
df.info()
df.shape
df.describe(include='all')
```

## 4. Check Missing Values and Duplicates

```python
df.isnull().sum()
df.duplicated().sum()
```

## 5. Inspect Unique Values
This helps identify inconsistent entries like UNKNOWN, ERROR or N/A.

```python
for col in df.columns:
    print(f'\n{col}')
    print(df[col].unique())
```

## 6. Replace Common Invalid Values

Many datasets use placeholders instead of true missing values.
We replace them with `np.nan` so pandas recognizes them as missing.


```python
invalid_values = ['UNKNOWN','Unknown','ERROR','Error','N/A','NULL','null','?','']
df.replace(invalid_values, np.nan, inplace=True)
df.head()
```

## 7. Convert Data Types


```python
for c in df.columns:
    print(c, '=', df[c].dtype)

numeric = ['Quantity', 'Price Per Unit', 'Total Spent',]
for c in numeric:
    df[c] = pd.to_numeric(df[c])

df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

## 8. Handle Missing Values


```python
for col in df.columns:
    if df[col].dtype == 'object':
        if df[col].isna().any():
            df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        if df[col].isna().any():
            df[col].fillna(df[col].median(), inplace=True)

df.isnull().sum()
```

## 9. Remove Duplicates

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

## 10. Final Validation

```python
df.info()
df.describe(include='all')
df.head()
```

## 11. Save Clean Dataset

```python
df.to_csv('cleaned_dirty_cafe_sales.csv', index=False)
print('Dataset saved successfully.')
```
